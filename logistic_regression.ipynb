```python
## import some dependencies
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report
```

```python
# Load the data
data_train = pd.read_csv('Data/Water_Train_Data.txt', sep='\t')
data_cv = pd.read_csv('Data/Water_CV_Data.txt', sep='\t')
data_test = pd.read_csv('Data/Water_Test_Data.txt', sep='\t')

# Check missing entries -> No missing entries
#data_train.isnull().values.any()
#data_cv.isnull().values.any()
#data_test.isnull().values.any()

X_train = data_train.drop('is_safe', axis=1)
X_cv = data_cv.drop('is_safe', axis=1)
X_test = data_test.drop('is_safe', axis=1)

y_train = data_train['is_safe']
y_cv = data_cv['is_safe']
y_test = data_test['is_safe']
```

```python
## Train
model = LogisticRegression(max_iter=1000, random_state=42)
model.fit(X_train, y_train)
```

```python
y_cv_pred = model.predict(X_cv)
print(classification_report(y_cv, y_cv_pred))
## Normal Not Balanced
```

```python
## Train (Balanced)
model_balanced = LogisticRegression(max_iter=1000, random_state=42, class_weight="balanced")
model_balanced.fit(X_train, y_train)
```

```python
y_cv_pred_balanced = model_balanced.predict(X_cv)
print(classification_report(y_cv, y_cv_pred_balanced))
## Normal Balanced
```

```python
# Use another Linear Regression model/classifier
from sklearn.linear_model import SGDClassifier
sgd_clf = SGDClassifier(loss='log_loss', penalty='l2', max_iter=1000, random_state=42)
sgd_clf.fit(X_train, y_train)
```

```python
y_cv_pred_sgd = sgd_clf.predict(X_cv)
print(classification_report(y_cv, y_cv_pred_sgd))
## SGD Not Balanced
```

```python
sgd_clf_balanced = SGDClassifier(loss='log_loss', penalty='l2', max_iter=1000, random_state=42, class_weight="balanced")
sgd_clf_balanced.fit(X_train, y_train)
```

```python
y_cv_pred_sgd_balanced = sgd_clf_balanced.predict(X_cv)
print(classification_report(y_cv, y_cv_pred_sgd_balanced))
## SDG Balanced
```

```python
# Try removing insignificant feature, viruses

## Redo "Dataset Pre-processing"
# Remove "viruses" column
data_train_new = data_train.copy()
data_cv_new = data_cv.copy()
data_test_new = data_test.copy()

data_train_new.drop(columns=['viruses'], inplace=True)
data_cv_new.drop(columns=['viruses'], inplace=True)
data_test_new.drop(columns=['viruses'], inplace=True)

X_train_new = data_train_new.drop('is_safe', axis=1)
X_cv_new = data_cv_new.drop('is_safe', axis=1)
X_test_new = data_test_new.drop('is_safe', axis=1)

y_train_new = data_train_new['is_safe']
y_cv_new = data_cv_new['is_safe']
y_test_new = data_test_new['is_safe']
```

```python
## Train
model_new = LogisticRegression(max_iter=1000, random_state=42)
model_new.fit(X_train_new, y_train_new)
```

```python
y_cv_pred_new = model_new.predict(X_cv_new)
print(classification_report(y_cv_new, y_cv_pred_new))
```

```python
model_new_balanced = LogisticRegression(max_iter=1000, random_state=42, class_weight="balanced")
model_new_balanced.fit(X_train_new, y_train_new)
```

```python
y_cv_pred_new_balanced = model_new_balanced.predict(X_cv_new)
print(classification_report(y_cv_new, y_cv_pred_new_balanced))
```

```python
sgd_clf_new = SGDClassifier(loss='log_loss', penalty='l2', max_iter=1000, random_state=42)
sgd_clf_new.fit(X_train_new, y_train_new)
```

```python
y_cv_pred_sgd_new = sgd_clf_new.predict(X_cv_new)
print(classification_report(y_cv_new, y_cv_pred_new))
```

```python
sgd_clf_new_balanced = SGDClassifier(loss='log_loss', penalty='l2', max_iter=1000, random_state=42, class_weight="balanced")
sgd_clf_new_balanced.fit(X_train_new, y_train_new)
```

```python
y_cv_pred_sgd_new_balanced = sgd_clf_new_balanced.predict(X_cv_new)
print(classification_report(y_cv_new, y_cv_pred_sgd_new_balanced))
```

Based on macro average F1-score, Logistic Regression using SGD classifier (balanced and without excluding "viruses")

```python
## Validate the best model with the test set
y_test_pred_sgd_balanced = sgd_clf_balanced.predict(X_test)
print(classification_report(y_test, y_test_pred_sgd_balanced))
```
